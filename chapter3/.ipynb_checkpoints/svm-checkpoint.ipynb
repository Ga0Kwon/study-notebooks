```python
#3-6 라이브러리 호출
from sklearn import svm
from sklearn import metrics
from sklearn import datasets
from sklearn import model_selection
import os 
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3' #TF_CPP_MIN_LOG_LEVEL이라는 환경변수를 사용하여 로깅 제어
# 0 : 모든 로그 표시
# 1 :INFO 로그 표시
# 2 : WARNING 로그 표시
# 3 : ERROR 로그 표시 
```

```python
#3-7 iris 데이터 준비 후 훈련과 테스트 데이터셋 분류
iris = datasets.load_iris() #사이킷런에서 제공하는 iris 데이터 호출
#사이킷런의 model_selection 패키지에서 제공하는 train_test_split 메서즈를 활요하여 훈련과 테스트 데이터셋 분리
#test_size = 0.6 (60%의 테스트 데이터), 훈련 데이터는 40%
#random_state = 42 데이터 섞는 방식을 고정 → 매번 같은 분할, 아무 숫자나 넣어도 됨 (42는 의미가 특별한 건 X)
X_train, X_test, y_train, y_test = model_selection.train_test_split(iris.data, iris.target, test_size=0.6, random_state=42)
```

```python
#3-8 SVM 모델에 대한 정확도

# SVM은 선형 분류와 비선형 분류(저차원 -> 고차원)를 지원
# 비선형 커널 등장 : 선형 분류될 수 없는 데이터로 인해 발생 -> 많은 수학적 계산으로 인해 성능에 문제를 줄 수 있음
svm = svm.SVC(kernel='linear', C=0.1, gamma=0.5)
#C: 오류 허용도. 크면 다 맞추려다 과적합, 작으면 너그러워 일반화. (모든 커널)
#gamma: 점의 영향 범위. 크면 좁아져 경계가 잘게 휘어 과적합, 작으면 넓어져 매끈. (RBF·다항식만)

svm.fit(X_train, y_train) #훈련 데이터를 사용하여 SVM 분류기를 훈련
perdictions = svm.predict(X_test) #훈련된 모델을 사용하여 테스트 데이터에서 예측
score = metrics.accuracy_score(y_test, perdictions) #정확도 계산
print('정확도 : {0:f}'.format(score))
```
